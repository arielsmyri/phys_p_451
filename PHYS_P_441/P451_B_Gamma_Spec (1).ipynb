**Introduction:**

---



In elementary modern physics, you learned of several different ways in which quanta of electromagnetic energy, or photons,
can interact with matter. The three dominant processes include: the photoelectric effect, in which all of the photon's
energy is transferred to a single electron; Compton scattering, in which the photon scatters from an electron, transferring
some of its energy, but re-emerging as a lower-energy photon; and pair production, whereby the photon converts to a  positron-electron pair.  In this lab, you will investigate processes involving high-energy photons, or gamma-rays, where, in this context, high-energy means on the scale of nuclear energy level spacings (i.e.,  several hundred keV to a few MeV).  Note that an MeV is many thousands of times larger than typical atomic energies, so the fact that the struck electrons are bound in matter is usually not important (but there are exceptions; if you are interested in delving deeper into this, you might explore the physics of "escape peaks" that sometimes show up in spectra).  These processes are described in more detail in [2,3].

The key tools in this experiment will be a Thallium-doped Sodium Iodide crystal, which is a scintillating crystal  often used as an ``electromagnetic calorimeter".  Nominally, the light produced by the crystal is proportional to the  gamma energy the crystal absorbs. Some fraction of this light will be collected with a Photomultiplier Tube (PMT),
which will convert that to a pulse of electrons (with total charge proportional to the light absorbed by the photo-cathode),
amplified (and differentiated) by a spectroscopy amplifier. The height of the resulting voltage pulse is therefore
proportional to the energy of the absorbed gamma ray, and the spectrum of photon energies is transformed into a histogram
by the multi-channel analyzer.  More information about photon detection and multi-channel analyzers is available in [2].

**Data read routine for Maestro MCB data**

```python

# This routine reads data (from "file:") that has been saved using the
#"File/Print to File" option to save the MCB BUFFER (!!) in Ortec's Maestro
# software). The routine assumes the data file is in the same director as this
#notebook
def read_maestro_text(file, calibration=(1,1)):
  ''' Read a spectrum from the Maestro software saved as text

  Parameters
  ----------
  file : string
    Text file to open
  calibration : {float,float}
    A tuple with the channel number and energy (keV) of the calibration peak

  Returns
  -------
    pandas DataFrame with columns:
      c    : channel number
      E    : calibrated energy of this channel in keV
      dE   : channel width in keV
      n    : counts recorded in this channel
      sign : statistical error on n computed as sqrt(n+1)

  '''
  #
  # The text files produced from Maestro Print->Print to file are formatted
  # following this example:0
  #
  # Line 1: <empty>
  # Line 2: Maestro spectrum printed on 1/13/2023 at 10:25:18 AM
  # Line 3: >BL-PHYS-904D9T2 Model 928 SN 14126199
  # Line 3:     0:          0         0         0         0         0         0         0
  # Line 4:     7:          0         0         0         0         0         0         0
  # Line 5:    14:          0         0         0         2         6         1         9
  #
  # and so on. These are the channel number, followed by the counts for the
  # channels starting at that numner
  #
  with open(file) as f:
    channel = list()
    counts  = list()
    iline = 0
    for line in f.readlines():
      if iline==0:   # No operation for leading blank line
        pass
      elif iline==1: # Header line. Print so users gets feedback
        print('Reading file for %s'%line)
      elif iline==2: # Another header. Skip this one.
        pass
      else:          # All other lines contain data
        n = line.replace(':',' ').split()
        for j in range(1,len(n)):
          #
          # j will range over the fields to the right of the ':'. Convert the
          # fields from strings to ints and assign channel and counts
          #
          ch = int(n[0])+j-1
          ct = int(n[0+j])
          #
          # Append these to the arrays. This may not be the most efficient.
          # Arrays want to be continuous in memory so the "appends" may require
          # them to be reallocated. However, we won't be executing this function
          # very often
          #
          channel.append(ch)
          counts. append(ct)
      iline = iline+1
    #
    # Compute errors on channel counts as sqrt(n+1) which works for both large
    # and small n's. If we've been provided an energy calibration, apply that
    # as well.
    #
    import numpy as np
    sign = np.sqrt(np.array(counts)+1)
    #if calibration[0]<=0.0:
    #  print('Bad calibration peak (%f) : Bailing',calibration[0])
    #  return None
    #dE = calibration[1]/calibration[0]
    #E  = dE*np.array(channel) This was Mark's calibration, let's get more
    # sophisticated. Default changed from (1,1) to (0,1) (DVB Aug. 2025)
    dE =1 # legacy to take care of residual references?
    E = calibration[0]+calibration[1]*np.array(channel)

    #
    # Package the data as a pandas dataframe for return.
    #
    import pandas as pd
    df = pd.DataFrame(data={'c':channel,
                            'E':E,
                            'dE':dE,
                            'n':counts,
                            'sig':sign})
    return df
#...............................................................................
def test_read_maestro_text(file):
  ''' Unit test for function read_maestro_text() '''
  df = read_maestro_text(file,calibration=(3000,622))
  df.hist()

  import matplotlib.pyplot as plt
  plt.show()

  plt.plot(df['c'],df['n'])
  plt.show()

  plt.errorbar(df['E'],df['n'],yerr=df['sig'],fmt='.')
  plt.show()
# test_read_maestro_text('cs137-13jan2023-run1.Txt')
```

**Routines for fitting to Gaussian peaks** (one or two) using nonlinear least squares fitting (Scipy routine curve_fit). Nonlinear fit routines require an initial guess for each adjustable parameter. For the single peak routine, reasonable guesses are made by the calling routine based on the global properties within the region of interest (ROI), for the two peak routine, the user mush provide some (use the plot to guide you).  

```python
# routines for fitting to a single Gaussian peak; starting guesses are
# provided within the fit_peak routine (based on global properties of the
# data, they may not work in all circumstances, but the do a reasonable job
# for most cases; do you understand how they are determined?)
import numpy as np
def peak_fcn(x, ab, mb, ap, xp, sigp):
  '''A function to represent a gaussian peak on a linear background

  Parameters
  ----------
  x - float
    the channel or energy we want to evaluate the functino for
  ab - float
    intercept of background
  mb - float
    slope of background
  ap - float
    normalization of gaussian peak
  xp - float
    peak location
  sigp - float
    peak width
  '''
  return ab+mb*x + ap*np.exp(-0.5*((x-xp)/sigp)**2)
#...............................................................................

def fit_peak(df, roi=(0,999999), mode='raw', draw=False):
  ''' Fit a peak to a particular region of interest
  Parameters
  ----------
  df : DataFrame
    The DataFrame for the spectrum we want to fit
  roi : tuple, optional
    The left and right edges of the region of interest we want to fit
  display : bool
    Display the results?

  Returns
  -------
    tuple of floats (peak, sigma peak, width, sigma width)
  '''
  ds = df.loc[(df["c"]>=roi[0]) & (df["c"]<=roi[1])]
  #
  # Provide some seeds for the fit. Asusme that the peak is roughly in the
  # center of the region of interest, has a width which isn't too different
  # from the width of the region of interest and is about as high as the
  # highest bin count insider the region of interest.
  # Seed the background at 1/10th the peak and have it fall off like
  # 1/(width of the region of interest)
  #
  from scipy.optimize import curve_fit

  xp    = 0.5*(roi[0]+roi[1])
  ap    = df["n"].max()
  sigp  = 0.5*(roi[1]-roi[0])
  # DVB the initial gueses below are for debugging, I need to
  # come up with a universal guess (how to specify E[roi[0]]??)
  if mode == 'cal': xp = 0.5*(ds["E"].min() + ds["E"].max())
  if mode == 'cal': ap = df["n"].max()
  if mode == 'cal': sigp = 0.5*(ds["E"].max()-ds["E"].min())
  ab    = 0.1*ap
  mb    = 0.0
  guess = [ab, mb, ap, xp, sigp]

  if mode == 'raw': x= ds["c"]
  if mode == 'cal': x= ds["E"]
  y= ds["n"]
  s= ds["sig"]

  (popt,pcov) = curve_fit(peak_fcn, x, y, sigma=s, p0=guess, absolute_sigma=True)
  perr        = np.sqrt(np.diag(pcov))
  best_fit_x  = x
  best_fit_y  = peak_fcn(best_fit_x,*popt)
  chisq       = np.sum(((y-best_fit_y)/s)**2)
  ndof        = len(best_fit_x)-5

  print('Best fit: chisq=%f / %d DOF'%(chisq, ndof))
  print(' intercept = %f +/- %f'%(popt[0],perr[0]))
  print(' slope = %f +/- %f'%(popt[1],perr[1]))
  print(' peak height = %f +/- %f'%(popt[2],perr[2]))
  print(' peak center = %f +/- %f'%(popt[3],perr[3]))
  print(' peak width  = %f +/- %f'%(popt[4],perr[4]))

  if draw:
    if mode == 'cal': draw_spectrum_calibrated(ds)
    if mode == 'raw': draw_spectrum_raw(ds)
    import matplotlib.pyplot as plt
    plt.plot(best_fit_x, best_fit_y, 'r-', zorder=100)

  return (popt[3], perr[3], popt[4], perr[4])

```

The task of finding an optimal (polynomial) calibration curve from the measured positions of known lines you perfomed above can be cast as a system of deg+1 coupled linear equations (wher deg is the degree of the polynomial you are fitting to) for which there is a unique solution (barring any degeneracies that render the system singular). This is an example of a "linear" least-squares problem, so we introduce below a different python fitting routine than the one we used for fitting the gaussians to your raw data.  We'll keep it simple for now and use a linear calibration curve (deg=1).

```python
# Use a linear least squares fit to calibrate the energy scale.
# x will be the known) Energy of each peak, y the fitted positino (in channels)
# and sigy the uncertainties on those channel positions (thiw may be different
# from what the fitting program suggests, Why?)

import numpy as np


# x-values are the known positions of lines for Na-22 and Co60 taken
# from the Lara Web site. y and sigy arrays should be replaced with your results

x = np.array([511.00,1173.228,1274.537,1332.492])
y = np.array([169.2,379.65,411.55,430.80])
sigy = np.array([0.05, 0.15, 0.1, 0.1])

# Weights for fitting are defined as (1 / sigma) for polyfit
weights = 1.0 / sigy

# Fit a 1st-degree polynomial (deg=1; straight line: y = mx + b);
# return parameter values to the array coeffs and provide the covariance matrix
# so that we can estimate the uncertainties on those coefficients.

coeffs, covariance_mat = np.polyfit(x, y, deg=1, w=weights, cov=True)

# Next we compute the fit at each data x-value and compute the
# Chi-squared value to assess the quality of the fit and check for blunders

y_fit=np.polyval(coeffs,x)
chi_sq = np.sum(((y-y_fit)/sigy)**2)

# Now, use the covariance matrix to determine the uncertainties on the parameters

uncerts=np.sqrt(np.diag(covariance_mat))
# Print a summary of the results:
print(f"       chi_sq= {chi_sq:.2f} for {len(x)-2:.0f} d.o.f.")
print(f"    Slope (m): {coeffs[0]:.4e} +/- {uncerts[0]:.1e}  chnl/keV")
print(f"Intercept (b): {coeffs[1]:.4e} +/-{uncerts[1]:.1e} chnl")

import matplotlib.pyplot as plt
plt.figure(1)
plt.errorbar(x,y,sigy, fmt='.')
plt.plot(x,y_fit)
#below we plot the normalized residuals of the fit

sigmamax=np.max(np.abs(sigy))
signrm=sigy/sigmamax

plt.figure(2)
plt.errorbar(x,(y-y_fit)/sigy, signrm, fmt='*')
#plt.plot(x,(y-y_fit)/sigy, fmt='*')
```
