**Introduction:**

---

Perhaps the two most fundamental measurements to be made in Condensed Matter Physics (CMP) are determining the longitudinal and transverse voltages that show up when an electrical current passes through a material (in finite or zero magnetic field). The "longitudinal" voltage ($V_{xx}$ is directly related to the sample's resistance (and therefore, through geometry, the material's resistivity). The transverse voltage ($V_{xy}$) is related to the Hall effect, which is a transverse electric field that builds up to counter-act the Lorentz force when a current flows through a magnetic field in a confined geometry (but see [1] for a remarkable surprise discovered in the 1980's that demonstrated it can be much more interesting than this under the right circumstances).

The material's resistivity can depend on many things (band structure, purity, grain structure, strain, temperature, magnetic field, ...). The transverse voltage is, in most cases, directly related to the density and charge of the mobile particles carrying current in the material (the "carrier" density) and the strength and direction of the magnetic field. Since both of these voltages are directly proportional to the current (in most circumstances; but consider reading up on cases where this is not the case [2,3]) we often talk of the longitudinal and transverse resistances ($R_{xx}$ and $R_{xy}$) in CMP.

Often physicists are most interested in $R_{xx}$ and $R_{xy}$ at low temperature (where effects such as lattice vibrations and smearing of the Fermi surface are less important, and subtle aspects of the band structure, disorder, and/or many body physics effects can be important). Making sure that the electrons are at the same temperature as your thermometer (or in some cases even reaching the temperature you are after, if that is below 100mK) may require the use of very small currents (say $\mu$A or less), and therefore these measurements may often require measuring very small voltages ($nV$ to $\mu V$) with rather large lead resistances in the presence of large thermal gradients. For instance, in reaching temperatures below 500mK, heat loads from thermal conduction along wires can be a limiting factor, so often you make use of alloy wires rather than copper to limit this thermal load; this can put even greater demands on using even smaller current to avoid joule heating in the leads carrying current.

**4-point resistance measurement:**


---

You have undoubtedly measured resistances earlier in your Physics career with an Ohmmeter or Digital Multimeter (DMM). This is an example of a "2-point" resistance measurement, in which you measure both the voltage and current at the source and use Ohm's law to give the resistance (well, the meter does that internally). You may not have thought of it before, but in doing the measurement this way you are assuming that your sample has no problem accepting whatever current comes from the Ohmmeter (typically driven with a few hundred mV), and, although you may assume your measurement gives the resistance of the sample, it actually measures the resistance of the sample PLUS that of the leads connecting the sample to the ohmmeter AND the contacts between those leads and the sample. In some cases (such as checking on sensitive electronics or cryogenic thermometers), you might damage the device simply by measuring its resistance with an ohmmeter or the meter might provide enough current to alter that resistance, so it is important to be aware of more sophisticated techniques for deteriming resistances.

**Thermal and Parasitic Voltages:**

---

In the above discussion, we assumed that the voltage measured was simply that given by Ohm's law for the sample's resistance and input current. However, any real voltage measurement has other potential contributions.

Any time a wire moves in a magnetic field or a static wire loop sees a time varying magnetic field (say from the 60Hz line current), transient voltages may appear in your circuit. We tend to minimize these voltages by putting wires in twisted pairs, using coaxial cables, trying not to move wires during measurements, and not running wires through areas of large magnetic fields.

Similarly, any time two dissimilar metals are in contact in a circuit with a temperature gradient, thermal voltages will appear. If you're interested in small enough voltages, this can be a problem even for small temperature differences and wires that are nominally the same. Therefore, it is best practice to always take precautions to account for thermal voltages.

Thermal voltages are most easily mitigated by simply reversing the direction of the current and subtracting the voltage measured for current in one direction from the other (the voltage of interest changes sign, the thermal voltages don't, and therefore cancel out; this leaves a value equal to twice the voltage of interest).  Adding the two voltages can give you a measurement of the thermal voltage (again with a factor of 2 of course). Note that there are other thermal effects that can depend on the direction of the current (see, for instance [4]), so this provides another reason for limiting current magnitudes whenever possible.  For these reasons, among others, the most sensitive measurements are typically taken with an alternating current rather than direct current. The latter also provides additional opportunities for mitigating against unwanted noise signals, but that is a subject for another lab (see the lock-in lab).

The code below computes and plots the best fit line to a set of V_xy measurements (in mV) for various currents (in micro-A), with the assumption that the V_xy uncertainty dominates and is constant and rougly 2uV. Using the discussion on pages 4 and 5 of [3] you can use this slope to compute the carrier density. You need to be careful with your geometry and sign conventions to get the carrier sign correct. In August of 2026, the magnetic fields were measured to be +/-0.727(2) T (in the directions indicated by the arrows). You should confirm that these have not changed.

```python
# Use a linear least squares fit to determine the Hall  coefficient from the
# measusred V_xy and I.

import numpy as np

# Hall effect data  x=Current (uA)  y=V_xy(mV)
x=np.array([16.47,36.239,149.31,105.256,67.025])
y=np.array([-0.046,-0.097,-0.398,-0.282,-0.179])
sigy =np.array([0.002,0.002,0.002,0.002,0.002])
# Weights for fitting are defined as (1 / sigma) for polyfit
weights = 1.0 / sigy

# Fit a 1st-degree polynomial (deg=1; straight line: y = mx + b);
# using a linear least-squares algorithm from numpy (poyfit).
# return parameter values to the array coeffs and provide the covariance matrix
# so that we can estimate the uncertainties on those coefficients.

coeffs, covariance_mat = np.polyfit(x, y, deg=1, w=weights, cov=True)

# Next we compute the fit at each data x-value and compute the
# Chi-squared value to assess the quality of the fit and check for blunders

y_fit=np.polyval(coeffs,x)
chi_sq = np.sum(((y-y_fit)/sigy)**2)

# Now, use the covariance matrix to determine the uncertainties on the parameters

uncerts=np.sqrt(np.diag(covariance_mat))
# Print a summary of the results:
print(f"       chi_sq= {chi_sq:.2f} for {len(x)-2:.0f} d.o.f.")
print(f"    Slope (m): {1000*coeffs[0]:.4e} +/- {uncerts[0]:.1e}  Ohms")
print(f"Intercept (b): {coeffs[1]:.4e} +/-{uncerts[1]:.1e} mV")

import matplotlib.pyplot as plt
plt.figure(1)
plt.errorbar(x,y,sigy, fmt='.')
plt.plot(x,y_fit)
#below we plot the normalized residuals of the fit

sigmamax=np.max(np.abs(sigy))
signrm=sigy/sigmamax

print(f"First plot is data with fit, second is the normalized residual trend")
plt.figure(2)
plt.errorbar(x,(y-y_fit)/sigy, signrm, fmt='*')

```
