**Summary:**

---
**Physics:** Nuclear decay can often involve the emission of multiple gamma rays. Measuring the angular correlations among those gammas can tell you a lot about the quantum mechanics of the system undergoing decay.  **Applications:** Corrleations between emitted gammas is central the Positron Emission Tomography (PET) medical imaging technique and angular correlations were also central to the 1975 discovery of "jets" in hadron production from e+/e- collisions (by former IU faculty member Gail Hanson when she as a post-doc at SLAC and for which she shared the 1996 Panofsky prize from the APS).  **Techniques:** You will learn about the power of coincidence counting techniques (e.g. for enhancing signal to background ratios) as well as how the details of a nuclear counting chain's setup can influence your measurements.


**Introduction**

---

Both $^{22}$Na and $^{60}$Co produce (nearly) coincident photons; however, the physics processes governing these two productions are entirely different.  In the case of $^{22}$Na the photons are produced by annihilation of a $\beta^+$ emitted from the isotope.  The process is $\beta^+ + \beta^- \to \gamma + \gamma$, where the $\beta^-$ is simply an electron in the matter surrounding the nucleus.  The two photons emerge simultaneously and in opposite directions (assuming the initial positronium is at rest).  One can verify this with the apparatus.  The $^{60}$Co nucleus undergoes $\beta$-decay to $^{60}$Ni.  This unstable Ni nucleus then decays through two sequential photon emissions.  (See Ref. [2,3,4] for more details).  The intermediate nuclear isotope has an extremely short half life on the order of picoseconds; therefore, these two photons are emitted nearly simultaneously.  Since the half-life of the intermediate state is so short, the polarization of the nucleus is preserved from the time the first photon is emitted until the second photon is emitted.  Therefore information about the angular momentum of the unstable nucleus is imprinted on the angular distribution of the emitted photons.  A measurement of this angular distribution is the topic of this lab.
